Qual estrutura de dados estudadas é mais adequada para armazenar os chamados aguardando atendimento?<br>
R: A estrutura mais adequada é o FIFO (first in first out) pois embora haja prioridade como caracteristica do chamado, a regra do negocios prioriza a ordem de **chegada**, portanto deve-se usar uma fila.<br>

---

Qual estrutura é mais adequada para guardar os chamados já atendidos, considerando que o sistema precisa desfazer primeiro o atendimento mais recente?<br>
R: A estrutura mais adequada para essa aplicação de negocio é o LIFO (Last in first out), pois eu preciso busca o ultimo atendimento realizado (o ultimo da lista) tendo o comportamento de pilha.<br>

---
Ordem de chegada
```
101 - Ana - Média
102 - Bruno - Alta
103 - Carla - Baixa
104 - Daniel - Alta
```
Sequência

```
atender
atender
desfazer
atender
```

R: As duas estruturas ficarão assim no final<br>
Atendidos
```
101 - Ana - Média
102 - Bruno - Alta
```
Espera

```
103 - Carla - Baixa
104 - Daniel - Alta
```



````python
class CentralAtendimento:
  def __init__(self):
    self.__chamado = []
    self.__atendido = []

  def adicionar_chamado(self, codigo, cliente, prioridade):
    self.__chamado.append([codigo, cliente, prioridade])

  def atender_proximo(self):
    self.__atendido.append(self.__chamado.pop(0))

  def desfazer_ultimo_atendimento(self):
    self.__chamado.insert(0, self.__atendido.pop())

  def buscar_chamado(self, codigo):
    for c in self.__chamado:
      if c[0] == codigo:
        return c
    return "Não encontrado"

  def exibir_situacao(self):
    print("-----Atendidos-----")
    print(self.__atendido)
    print("-----Aguardando-----")
    print(self.__chamado)
````

````python
central = CentralAtendimento()

central.adicionar_chamado(101, "Ana", "Média")
central.adicionar_chamado(102, "Bruno", "Alta")
central.adicionar_chamado(103, "Carla", "Baixa")
central.adicionar_chamado(104, "Daniel", "Alta")

central.atender_proximo()
central.atender_proximo()

central.desfazer_ultimo_atendimento()

central.atender_proximo()

central.exibir_situacao()
````

````python
print(central.buscar_chamado(104))
````

A complexidade do metodo buscar_chamado() é O(n) pois no pior cenário deverá percorrer toda a lista.
